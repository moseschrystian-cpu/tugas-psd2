# **CHI-SQUARE DISTRIBUTION**

Probability Distribution Function (PDF)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

```python
df = 5
n = 100
```

```python
data = np.random.chisquare(df, n)
```

```python
print(data)
```

```python
tabel = pd.DataFrame({
    "No": range(1, n + 1),
    "Data Chi-Square": data
})

print(tabel)
```

```python
x = np.linspace(0, max(data), 1000)
pdf = stats.chi2.pdf(x, df)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(x, pdf, label="PDF Chi-Square")

plt.title("Probability Distribution - Chi-Square")
plt.xlabel("X")
plt.ylabel("Probability Density")

plt.grid()
plt.legend()
plt.show()
```

Cumulative Distribution Function (CDF)

```python
cdf = stats.chi2.cdf(x, df)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(x, cdf, label="CDF Chi-Square")

plt.title("CDF - Chi-Square Distribution")
plt.xlabel("X")
plt.ylabel("Cumulative Probability")

plt.grid()
plt.legend()
plt.show()
```

HISTOGRAM CHI-SQUARE DISTRIBUTION

```python
plt.figure(figsize=(8,5))

plt.hist(data, bins=15, edgecolor='black')

plt.title("Histogram Data Chi-Square")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")

plt.show()
```

# **F-DISTRIBUTION**

Probability Distribution Function (PDF)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

```python
df1 = 5
df2 = 10
n = 100
```

```python
data = np.random.f(df1, df2, n)
```

```python
print(data[:10])
```

```python
tabel = pd.DataFrame({
    "No": range(1, n + 1),
    "Data F-Distribution": data
})

print(tabel)
```

```python
x = np.linspace(0, max(data), 1000)
pdf = stats.f.pdf(x, df1, df2)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(x, pdf, label="PDF F-Distribution")

plt.title("Probability Distribution - F-Distribution")
plt.xlabel("X")
plt.ylabel("Probability Density")

plt.grid()
plt.legend()
plt.show()
```

Cumulative Distribution Function (CDF)

```python
cdf = stats.f.cdf(x, df1, df2)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(x, cdf, label="CDF F-Distribution")

plt.title("CDF - F-Distribution")
plt.xlabel("X")
plt.ylabel("Cumulative Probability")

plt.grid()
plt.legend()
plt.show()
```

HISTOGRAM F-DISTRIBUTION

```python
plt.figure(figsize=(8,5))

plt.hist(data, bins=15, edgecolor='black')

plt.title("Histogram Data F-Distribution")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")

plt.show()
```
